```python
import gymnasium as gym
import numpy as np
import matplotlib.pyplot as plt
```

```python
# -------------------------------------------------
# Create FrozenLake Environment
# -------------------------------------------------
env_desc = [
    "SFFF",
    "FHFH",
    "FFFH",
    "HFFG"
]

env = gym.make("FrozenLake-v1", desc=env_desc, is_slippery=True)
# write your code here
```

```python
# -------------------------------------------------
# Value Iteration Algorithm
# -------------------------------------------------

def value_iteration(env, gamma=0.99, theta=1e-8):
    """
    Performs Value Iteration and returns:
    - Optimal state-value function V
    - Optimal policy
    - Number of iterations
    """

    # Access the underlying environment
    model = env.unwrapped

    # Number of states and actions
    n_states = model.observation_space.n
    n_actions = model.action_space.n

    # Initialize value function
    V = np.zeros(n_states)

    # Iteration counter
    iteration = 0

    while True:

        # Maximum change in value function
        delta = 0

        # Create a copy so that all states
        # use values from the previous iteration
        V_new = np.copy(V)

        # Update every state
        for state in range(n_states):

            # Calculate Q-value for every action
            action_values = []

            for action in range(n_actions):

                q_value = 0

                # P[state][action] contains:
                # probability, next_state, reward, terminated
                for probability, next_state, reward, terminated in model.P[state][action]:

                    q_value += probability * (
                        reward + gamma * V[next_state] * (not terminated)
                    )

                action_values.append(q_value)

            # Bellman Optimality Equation
            V_new[state] = max(action_values)

            # Calculate maximum change
            delta = max(delta, abs(V_new[state] - V[state]))

        # Update value function
        V = V_new

        iteration += 1

        # Check convergence
        if delta < theta:
            break

    # -------------------------------------------------
    # Extract Optimal Policy
    # -------------------------------------------------

    policy = np.zeros(n_states, dtype=int)

    for state in range(n_states):

        action_values = []

        for action in range(n_actions):

            q_value = 0

            for probability, next_state, reward, terminated in model.P[state][action]:

                q_value += probability * (
                    reward + gamma * V[next_state] * (not terminated)
                )

            action_values.append(q_value)

        # Select action with maximum Q-value
        policy[state] = np.argmax(action_values)

    return V, policy, iteration
```

```python
# -------------------------------------------------
# Run Value Iteration
# -------------------------------------------------

# write your code here
V, policy, iterations = value_iteration(
    env,
    gamma=0.99,
    theta=1e-8
)
```
